```python
import pm4py
import pandas as pd
# load data
df1 = pd.read_csv('data/bookstore.csv', sep=',')
df2 = pd.read_csv('data/partiallyOrdered.csv', sep=',')
df3 = pd.read_csv('data/processComplaints.csv', sep=',')
df4 = pd.read_csv('data/repair.csv', sep=',')
df5 = pd.read_csv('data/travel.csv', sep=',')
print('##### bookstore ####')
print(df1)
print('##### partiallyOrdered ####')
print(df2)
print('##### processComplaints ####')
print(df3)
print('##### repair ####')
print(df4)
print('##### travel ####')
print(df5)
```

```python
# number of cases 
print ('le nombre de cases de bookstore : ' + str(len(df1['case:concept:name'].unique())))
print ('le nombre de cases de partially ordered : ' + str(len(df2['case:concept:name'].unique())))
print ('le nombre de cases de process Complaints : ' + str(len(df3['case:concept:name'].unique())))
print ('le nombre de cases de repair : ' + str(len(df4['case:concept:name'].unique())))
print ('le nombre de cases de travel : ' + str(len(df5['case:concept:name'].unique())))
```

```python
# load logs 
log1 = pm4py.format_dataframe(df1, case_id='case:concept:name',activity_key='concept:name',timestamp_key='time:timestamp')
log2 = pm4py.format_dataframe(df2, case_id='case:concept:name',activity_key='concept:name',timestamp_key='time:timestamp')
log3 = pm4py.format_dataframe(df3, case_id='case:concept:name',activity_key='concept:name',timestamp_key='time:timestamp')
log4 = pm4py.format_dataframe(df4, case_id='case:concept:name',activity_key='concept:name',timestamp_key='time:timestamp')
log5 = pm4py.format_dataframe(df5, case_id='case:concept:name',activity_key='concept:name',timestamp_key='time:timestamp')

print('##### bookstore ####')
print(log1)
print('##### partiallyOrdered ####')
print(log2)
print('##### processComplaints ####')
print(log3)
print('##### repair ####')
print(log4)
print('##### travel ####')
print(log5)
```

```python
# number of variants
variants1 = pm4py.get_variants_as_tuples(log1)
print ('le nombre de variants de bookstore : ' + str(len(variants1)))
variants2 = pm4py.get_variants_as_tuples(log2)
print ('le nombre de cases de partially ordered : ' + str(len(variants2)))
variants3 = pm4py.get_variants_as_tuples(log3)
print ('le nombre de cases de process Complaints : ' + str(len(variants3)))
variants4 = pm4py.get_variants_as_tuples(log4)
print ('le nombre de cases de repair : ' + str(len(variants4)))
variants5 = pm4py.get_variants_as_tuples(log5)
print ('le nombre de cases de travel : ' + str(len(variants5)))
```

```python
# alpha algorithm
print ('book store')
net1, initial_marking1, final_marking1 = pm4py.discover_petri_net_alpha(log1)
pm4py.view_petri_net(net1, initial_marking1, final_marking1)
print ('partiallyOrdered')
net2, initial_marking2, final_marking2 = pm4py.discover_petri_net_alpha(log2)
pm4py.view_petri_net(net2, initial_marking2, final_marking2)
print ('process complaints')
net3, initial_marking3, final_marking3 = pm4py.discover_petri_net_alpha(log3)
pm4py.view_petri_net(net3, initial_marking3, final_marking3)
print ('repair')
net4, initial_marking4, final_marking4 = pm4py.discover_petri_net_alpha(log4)
pm4py.view_petri_net(net4, initial_marking4, final_marking4)
print ('travel')
net5, initial_marking5, final_marking5 = pm4py.discover_petri_net_alpha(log5)
pm4py.view_petri_net(net5, initial_marking5, final_marking5)
```

```python
# quality criteria : alpha algorithm

from pm4py.algo.evaluation.generalization import algorithm as generalization_evaluator
from pm4py.algo.evaluation.simplicity import algorithm as simplicity_evaluator

print ('####### book store')
fitness_token = pm4py.fitness_token_based_replay(log1, net1, initial_marking1, final_marking1)
fitness_alignment = pm4py.fitness_alignments(log1, net1, initial_marking1, initial_marking1)
prec_token = pm4py.precision_token_based_replay(log1, net1, initial_marking1, final_marking1)
#prec_alignment = pm4py.precision_alignments(log1, net1, initial_marking1, final_marking1)
gen = generalization_evaluator.apply(log1, net1, initial_marking1, final_marking1)
simp = simplicity_evaluator.apply(net1)

print('Fitness token based', fitness_token)
print('Fitness alignments', fitness_alignment)
print('Precision token based', prec_token)
# print('Precision alignments', prec_alignment)
print('Generalization', gen)
print('Simplicity', simp)
print('Soundness', pm4py.check_soundness(net1, initial_marking1,final_marking1)[0])

print ('####### partiallyOrdered')
fitness_token = pm4py.fitness_token_based_replay(log2, net2, initial_marking2, final_marking2)
#fitness_alignment = pm4py.fitness_alignments(log2, net2, initial_marking2, final_marking2)
prec_token = pm4py.precision_token_based_replay(log2, net2, initial_marking2, final_marking2)
#prec_alignment = pm4py.precision_alignments(log2, net2, initial_marking2, final_marking2)
gen = generalization_evaluator.apply(log2, net2, initial_marking2, final_marking2)
simp = simplicity_evaluator.apply(net2)

print('Fitness token based', fitness_token)
#print('Fitness alignments', fitness_alignment)
print('Precision token based', prec_token)
# print('Precision alignments', prec_alignment)
print('Generalization', gen)
print('Simplicity', simp)
print('Soundness', pm4py.check_soundness(net2, initial_marking2,final_marking2)[0])

print ('####### process complaints')
fitness_token = pm4py.fitness_token_based_replay(log3, net3, initial_marking3, final_marking3)
#fitness_alignment = pm4py.fitness_alignments(log3, net3, initial_marking3, final_marking3)
prec_token = pm4py.precision_token_based_replay(log3, net3, initial_marking3, final_marking3)
#prec_alignment = pm4py.precision_alignments(log1, net1, initial_marking1, final_marking1)
gen = generalization_evaluator.apply(log3, net3, initial_marking3, final_marking3)
simp = simplicity_evaluator.apply(net3)

print('Fitness token based', fitness_token)
#print('Fitness alignments', fitness_alignment)
print('Precision token based', prec_token)
# print('Precision alignments', prec_alignment)
print('Generalization', gen)
print('Simplicity', simp)
print('Soundness', pm4py.check_soundness(log3, net3, initial_marking3, final_marking3)[0])

print ('####### repair')
fitness_token = pm4py.fitness_token_based_replay(log4, net4, initial_marking4, final_marking4)
#fitness_alignment = pm4py.fitness_alignments(log4, net4, initial_marking4, final_marking4)
prec_token = pm4py.precision_token_based_replay(log4, net4, initial_marking4, final_marking4)
#prec_alignment = pm4py.precision_alignments(log4, net4, initial_marking4, final_marking4)
gen = generalization_evaluator.apply(log4, net4, initial_marking4, final_marking4)
simp = simplicity_evaluator.apply(net4)

print('Fitness token based', fitness_token)
#print('Fitness alignments', fitness_alignment)
print('Precision token based', prec_token)
# print('Precision alignments', prec_alignment)
print('Generalization', gen)
print('Simplicity', simp)
print('Soundness', pm4py.check_soundness(log4, net4, initial_marking4, final_marking4)[0])

print ('####### travel')
fitness_token = pm4py.fitness_token_based_replay(log5, net5, initial_marking5, final_marking5)
#fitness_alignment = pm4py.fitness_alignments(log5, net5, initial_marking5, initial_marking5)
prec_token = pm4py.precision_token_based_replay(log5, net5, initial_marking5, initial_marking5)
#prec_alignment = pm4py.precision_alignments(log5, net5, initial_marking5, initial_marking5)
gen = generalization_evaluator.apply(log5, net5, initial_marking5, initial_marking5)
simp = simplicity_evaluator.apply(net5)

print('Fitness token based', fitness_token)
#print('Fitness alignments', fitness_alignment)
print('Precision token based', prec_token)
# print('Precision alignments', prec_alignment)
print('Generalization', gen)
print('Simplicity', simp)
print('Soundness', pm4py.check_soundness(log5, net5, initial_marking5, initial_marking5)[0])
```
